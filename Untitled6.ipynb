```python
df_country=pd.read_excel('Country-Code.xlsx')  #it is a excel file so we need to use _excel
df_country.head()
```

```python
df.columns
```

```python
final_df=pd.merge(df,df_country,on='Country Code',how='left')
```

```python
final_df.head(2)
```

```python
final_df.dtypes  #to check datatypes
```

```python
final_df.columns
```

```python
final_df.Country.value_counts()   #how many diff cont and wrt to this country how many records are here
```

```python
Country_names=final_df.Country.value_counts().index
```

```python
Country_val=final_df.Country.value_counts().values
```

```python
#pie chart-top 3 countries that uses zomato
plt.pie(Country_val[:3],labels=Country_names[:3],autopct='%1.2f%%')
```

Observation: maximum records or transactions are from india
After that USA and UK

```python
#Numerical Variables
final_df.columns
```

```python
ratings=final_df.groupby(['Aggregate rating','Rating color', 'Rating text']).size().reset_index().rename(columns={0:'Rating Count'})
```

```python
ratings
```

```python
#conclusions
```

#conclusions  
4.5 to 4.9 excellent
4.0 t0 3.4 very good
3.5 to 3.9 good...

```python
ratings.head()
```

```python
import matplotlib
matplotlib.rcParams['figure.figsize']=(12,6)
sns.barplot(x='Aggregate rating',y='Rating Count',hue='Rating color',data=ratings,palette=['blue','red','yellow','green','green'])
```

observation
1.not rated is very high
2.maximum number of ratings are between 2.5 to 3.4

```python
##count plot
sns.countplot(x='Rating color',data=ratings,palette=['blue','red','yellow','green','green'])
```

```python
#create a pie chart for cities distributions
City_values=final_df.City.value_counts().values
City_labels=final_df.City.value_counts().index
```

```python
plt.pie(City_values[:5],labels=City_labels[:5],autopct='%1.2f%%')
```

```python
#find top 10 cuisines
# Find top 10 cuisines
top_cuisines = final_df['Cuisines'].value_counts().head(10)
top_cuisines
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))
sns.barplot(x=top_cuisines.values, y=top_cuisines.index, palette='magma')
plt.title('Top 10 Cuisines on Zomato')
plt.xlabel('Number of Restaurants')
plt.ylabel('Cuisines')
plt.show()
```
